```python
import math

# Classifier 1
print("Enter equation coefficients for Classifier 1")

a0 = float(input("a0 = "))
a1 = float(input("a1 = "))
a2 = float(input("a2 = "))

error1 = 2 / math.sqrt(a1**2 + a2**2)

# Classifier 2
print("\nEnter equation coefficients for Classifier 2")

b0 = float(input("a0 = "))
b1 = float(input("a1 = "))
b2 = float(input("a2 = "))

error2 = 2 / math.sqrt(b1**2 + b2**2)

# Display equations
print("\nClassifier 1:")
print(a0, "+", a1, "x1 +", a2, "x2 = 0")
print("Distance Error =", error1)

print("\nClassifier 2:")
print(b0, "+", b1, "x1 +", b2, "x2 = 0")
print("Distance Error =", error2)

# Compare
if error1 < error2:
    print("\nClassifier 1 is the good classifier")
elif error2 < error1:
    print("\nClassifier 2 is the good classifier")
else:
    print("\nBoth classifiers have same distance error")
```

```python
import numpy as np

# Positive class
S1 = np.array([3, 1])
S2 = np.array([3, 2])
S3 = np.array([4, 0])

# Negative class
S4 = np.array([1, 0])
S5 = np.array([0, 1])
S6 = np.array([0, -1])

# Augmented vectors (third coordinate = 1)
S1_aug = np.append(S1, 1)
S2_aug = np.append(S2, 1)
S3_aug = np.append(S3, 1)
S4_aug = np.append(S4, 1)
S5_aug = np.append(S5, 1)
S6_aug = np.append(S6, 1)

# Alpha values
alpha = [0.4, 0, 0, 0.4, 0, 0]

# Find w
w = (alpha[0] * S1) + (alpha[1] * S2) + (alpha[2] * S3) \
    - (alpha[3] * S4) - (alpha[4] * S5) - (alpha[5] * S6)

# Find b
b = 1 - np.dot(w, S1)

# Augmented weight vector
w_aug = np.append(w, b)

# Display results
print("Alpha values:")
for i in range(6):
    print("alpha" + str(i + 1) + " =", alpha[i])

print("\nWeight vector:")
print("w =", w)

print("\nw1 =", w[0])
print("w2 =", w[1])
print("b =", b)

print("\nAugmented weight vector:")
print("w_tilde =", w_aug)

print("\nOptimal Hyperplane:")
print(w[0], "x1 +", w[1], "x2 +", b, "= 0")

print("\nSimplified Hyperplane:")
print("4x1 + 2x2 - 9 = 0")
```
