```python
from scipy.stats import chi2_contingency
from scipy.stats import chi2
from scipy.stats import ttest_ind

table = [
    [35, 15],
    [25, 25]
]

print("\nObservation Table")
print(table)

stat, p, dof, expected = chi2_contingency(table)

print("\nExpectation Table")
print(expected)

prob = 0.90

critical = chi2.ppf(prob, dof)

print("\nCritical Value")
print(critical)

print("Statistical Value")
print(stat)

if abs(stat) >= critical:
    print("Stat value greater than or equal to critical: Reject Null Hypothesis")
else:
    print("Stat value less than critical: Accept Alternative Hypothesis")

x = [6, 3, 4, 5, 5, 9, 7, 8]
y = [4, 4, 1, 3, 5, 6, 2, 7]

print("\nt-Test Completed\n")

stat, p = ttest_ind(x, y, equal_var=True)

print("Statistical Value")
print(stat)

print("p Value")
print(p)
```
