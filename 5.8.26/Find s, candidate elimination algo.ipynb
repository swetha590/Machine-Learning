```python
import pandas as pd

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

df = pd.DataFrame(data, columns=['Sky', 'AirTemp', 'Humidity', 'Wind', 'Water', 'Forecast', 'Play'])

concepts = df.iloc[:, :-1].values
target = df.iloc[:, -1].values

hypothesis = None

for i in range(len(target)):
    if target[i] == 'Yes':
        if hypothesis is None:
            hypothesis = concepts[i].copy()
        else:
            for j in range(len(hypothesis)):
                if hypothesis[j] != concepts[i][j]:
                    hypothesis[j] = '?'

print("Final Hypothesis:")
print(hypothesis)
```

```python
import numpy as np

data = np.array([
    ['Sunny','Warm','Normal','Strong','Warm','Same','Yes'],
    ['Sunny','Warm','High','Strong','Warm','Same','Yes'],
    ['Rainy','Cold','High','Strong','Warm','Change','No'],
    ['Sunny','Warm','High','Strong','Cool','Change','Yes']
])

concepts = data[:, :-1]
target = data[:, -1]

def learn(concepts, target):
    specific = concepts[0].copy()
    general = [['?' for _ in range(len(specific))] for _ in range(len(specific))]

    print("Initialization of Specific Hypothesis")
    print(specific)

    print("\nInitialization of General Hypothesis")
    print(np.array(general))

    for i, h in enumerate(concepts):
        if target[i] == "Yes":
            for x in range(len(specific)):
                if h[x] != specific[x]:
                    specific[x] = '?'
                    general[x][x] = '?'
        else:
            for x in range(len(specific)):
                if h[x] != specific[x]:
                    general[x][x] = specific[x]
                else:
                    general[x][x] = '?'

        print("\nStep", i + 1)
        print("Specific:", specific)
        print("General:")
        print(np.array(general))

    general = [g for g in general if g != ['?'] * len(specific)]

    print("\nFinal Specific Hypothesis:")
    print(specific)

    print("\nFinal General Hypothesis:")
    for g in general:
        print(g)

learn(concepts, target)
```
